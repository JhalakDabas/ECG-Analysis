```python
# Denoising
from scipy import signal

def remove_motion_artifacts(raw_signal, fs=400):
    low_cutoff = 0.5
    high_cutoff = 45.0

    nyquist = 0.5 * fs
    low = low_cutoff / nyquist
    high = high_cutoff / nyquist

    b, a = signal.butter(4, [low, high], btype='band')

    filtered_signal = signal.filtfilt(b, a, raw_signal)

    return filtered_signal

clean_ecg_signal = remove_motion_artifacts(raw_ecg_signal, fs=400)

duration = 4000
time = [i/400 for i in range(duration)]

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(25, 7), sharex=True)
ax1.plot(time, raw_ecg_signal[:duration], color='crimson', label="Raw Corrupted ECG")
ax1.set_title("Before Preprocessing (Motion Contaminated)")
ax1.grid(True)

ax2.plot(time, clean_ecg_signal[:duration], color='teal', label="Clean ECG")
ax2.set_title("After Motion Artifact Removal")
ax2.set_xlabel("Time (Seconds)")
ax2.grid(True)
plt.show()
```

```python
signal_chunk = raw_ecg_signal
fs = 400
N = len(signal_chunk)

time_axis = np.arange(N) / fs

r_peaks, _ = signal.find_peaks(signal_chunk, distance=int(fs*0.6), prominence=0.2)
q_peaks = []
s_peaks = []
midpoints_x = []
midpoints_y = []

for r in r_peaks:
    window_q = signal_chunk[max(0, r - int(fs*0.1)):r]
    window_s = signal_chunk[r:min(N, r + int(fs*0.1))]

    if len(window_q) == 0 or len(window_s) == 0:
        continue

    q = max(0, r - int(fs*0.1)) + np.argmin(window_q)
    s = r + np.argmin(window_s)

    q_peaks.append(q)
    s_peaks.append(s)
    midpoints_x.append((q + s) / 2.0)
    midpoints_y.append((signal_chunk[q] + signal_chunk[s]) / 2.0)

plt.figure(figsize=(25, 5))
plt.plot(time_axis, signal_chunk, color='crimson', lw=1.2, label='Raw Corrupted Signal')

plt.scatter(np.array(r_peaks)/fs, signal_chunk[r_peaks], color='blue', marker='d', s=60, label='R-Peaks', zorder=5)
plt.scatter(np.array(q_peaks)/fs, signal_chunk[q_peaks], color='green', marker='s', s=40, label='Q-Peaks', zorder=5)
plt.scatter(np.array(s_peaks)/fs, signal_chunk[s_peaks], color='black', marker='s', s=40, label='S-Peaks', zorder=5)

plt.ylim(2, 4)
plt.xlim(0, 8)
plt.title("Identified Heartbeat Landmark Boundaries", fontsize=12, fontweight='bold')
plt.xlabel("Time (Seconds)")
plt.ylabel("Voltage (mV)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
from scipy.interpolate import CubicSpline

midpoints_x_padded = [0] + midpoints_x + [N - 1]
midpoints_y_padded = [midpoints_y[0]] + midpoints_y + [midpoints_y[-1]]

spline_model = CubicSpline(midpoints_x_padded, midpoints_y_padded)
estimated_baseline = spline_model(np.arange(N))

plt.figure(figsize=(25, 5))
plt.plot(time_axis, signal_chunk, color='crimson', alpha=1.0, label='Raw Corrupted Signal')

plt.scatter(np.array(r_peaks)/fs, signal_chunk[r_peaks], color='blue', marker='d', s=60, label='R-Peaks', zorder=5)
plt.scatter(np.array(q_peaks)/fs, signal_chunk[q_peaks], color='green', marker='s', s=40, label='Q-Peaks', zorder=5)
plt.scatter(np.array(s_peaks)/fs, signal_chunk[s_peaks], color='black', marker='s', s=40, label='S-Peaks', zorder=5)

plt.plot(time_axis, estimated_baseline, color='darkorange', lw=2.5, linestyle='--', label='Tracked Cubic Spline Baseline')

# plt.ylim(-0.3, 0.3)
plt.xlim(954,958)
plt.title("Low-Frequency Baseline Wander Tracking Curve", fontsize=12, fontweight='bold')
plt.xlabel("Time (Seconds)")
plt.ylabel("Voltage (mV)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

final_output_signal = np.copy(ecg_no_baseline)
standard_length = 250
window_size = 3
standardized_segments = []
sq_ranges = []

if len(s_peaks) > 1:
    for i in range(len(s_peaks) - 1):
        start_idx = s_peaks[i]
        end_idx = q_peaks[i+1]
        sq_ranges.append((start_idx, end_idx))

        seg = ecg_no_baseline[start_idx:end_idx]

        if len(seg) >= 5:
            x_old = np.linspace(0, 1, len(seg))
            x_new = np.linspace(0, 1, standard_length)
            standardized_segments.append(np.interp(x_new, x_old, seg))
        else:
            standardized_segments.append(np.zeros(standard_length))

    standardized_segments = np.array(standardized_segments)

    for i in range(len(standardized_segments)):
        start_w = max(0, i - window_size // 2)
        end_w = min(len(standardized_segments), i + window_size // 2 + 1)

        mean_segment = np.mean(standardized_segments[start_w:end_w], axis=0)

        start_idx, end_idx = sq_ranges[i]
        original_length = end_idx - start_idx
        x_old = np.linspace(0, 1, standard_length)
        x_new = np.linspace(0, 1, original_length)

        final_output_signal[start_idx:end_idx] = np.interp(x_new, x_old, mean_segment)

duration_samples = 2500
time_axis_6s = np.arange(duration_samples) / 400

plt.figure(figsize=(25, 6))
plt.plot(time_axis_6s, raw_ecg_signal[:duration_samples], color='crimson', alpha=0.7, linewidth=1.5, label='1. Original Raw Signal (Noisy)')
plt.plot(time_axis_6s, final_output_signal[:duration_samples], color='darkviolet', alpha=1.0, linewidth=1.5, label='2. Final Step 4 Output (QRSMR Cleaned)')
plt.title("Raw Signal vs. Fully Cleared Signal", fontsize=14, fontweight='bold')
plt.xlabel("Time (Seconds)", fontsize=12)
plt.ylabel("Voltage Amplitude (mV)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy import signal
import matplotlib.pyplot as plt

def subtraction_baseline_filter(raw_signal, fs=400):
    nyquist = 0.5 * fs
    cutoff = 0.5 / nyquist
    b, a = signal.butter(2, cutoff, btype='low')
    isolated_drift = signal.filtfilt(b, a, raw_signal)
    subtracted_clean_signal = raw_signal - isolated_drift
    return subtracted_clean_signal, isolated_drift

duration_samples = 2500
time_axis_6s = np.arange(duration_samples) / 400

subtracted_ecg, isolated_noise = subtraction_baseline_filter(raw_ecg_signal, fs=400)

plt.figure(figsize=(25, 6))
plt.plot(time_axis_6s, raw_ecg_signal[:duration_samples], color='crimson', alpha=0.6, linewidth=1.5, label='1. Original Raw Signal')
plt.plot(time_axis_6s, subtracted_ecg[:duration_samples], color='darkgreen', alpha=1.0, linewidth=1.5, label='2. Subtracted Filter Output')
plt.title("Motion Artifact Removal via Low-Frequency Noise Subtraction", fontsize=14, fontweight='bold')
plt.xlabel("Time (Seconds)", fontsize=12)
plt.ylabel("Voltage Amplitude (mV)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
import pywt
import numpy as np
import matplotlib.pyplot as plt

def wavelet_subtraction_filter(raw_signal, wavelet='sym4', level=5):
    coeffs = pywt.wavedec(raw_signal, wavelet, level=level)
    coeffs[0] = np.zeros_like(coeffs[0])

    sigma = np.median(np.abs(coeffs[-1])) / 0.6745
    threshold = sigma * np.sqrt(2 * np.log(len(raw_signal)))

    for i in range(1, len(coeffs)):
        coeffs[i] = pywt.threshold(coeffs[i], value=threshold, mode='soft')

    return pywt.waverec(coeffs, wavelet)[:len(raw_signal)]

duration_samples = 2500
time_axis_6s = np.arange(duration_samples) / 400

wavelet_subtracted_ecg = wavelet_subtraction_filter(raw_ecg_signal)

plt.figure(figsize=(25, 6))
plt.plot(time_axis_6s, raw_ecg_signal[:duration_samples], color='crimson', alpha=0.6, label='1. Original Raw Signal')
plt.plot(time_axis_6s, wavelet_subtracted_ecg[:duration_samples], color='teal', alpha=1.0, linewidth=1.5, label='2. Wavelet Cleaned Output')

plt.title("Motion Artifact Removal via Wavelet Subtraction (Fixed Array Formats)", fontsize=14, fontweight='bold')
plt.xlabel("Time (Seconds)")
plt.ylabel("Voltage Amplitude (mV)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
